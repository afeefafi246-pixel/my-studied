```python
letters=['a','b','d','e','i','j','o']     #filter function

def filter_vowels(letter):
  vowels=['a','e','i','o','u']
  if letter in vowels:
    return True
  else:
    return False

  filtered_vowels=tuple(filter(filter_vowels,letters))
  print(filtered_vowels)
```
