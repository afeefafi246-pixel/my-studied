```python
class employee:
  def __init__(self,name,emp_id,base_salary):
    self.name=name
    self.emp_id=emp_id
    self.base_salary=base_salary
  def show(self):
    print("name :",self.name)
    print("emp_id :",self.emp_id)
    print("base_salary :",self.base_salary)
class full_time(employee):
  def __init__(self,bonus):
    self.bonus=bonus
  def display(self):
    print("full time employee :",self.base_salary+self.bonus)
class part_time():
   def __init__(self,name,emp_id,work,rate):
    self.name=name
    self.emp_id=emp_id
    self.work=work
    self.rate=rate
   def see(self):
    print("name :",self.name)
    print("emp_id :",self.emp_id)
    print("work :",self.work)
    print("rate :",self.rate)
c=employee("afeef",102,6000)
f1=c.full_time(3000)
p1=part_time("sahal",201,3,3000)
print(f1.display())
```
