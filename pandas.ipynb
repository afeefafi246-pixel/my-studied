```python
DATA=[20,20,30,40,40,50,60]
arr=pd.Series(DATA)  #series was used for one line numbers
print(arr)
print(arr.unique())  #nique was used for delete the repeated numbers
print(arr.nunique()) #nunique is used for show how many unique numbers
print(arr.value_counts())  #value_counts is used for count how many times each unique value occures in a series
```

```python
DATA=[20,20,30,None,None,40,40,50,60]
arr=pd.Series(DATA)  #series was used for one line numbers
print(arr)
print(arr.isnull())  #isnull --- is show true on null value and show false on not null
print(arr.notnull())  #notnull -- is show true on not null values and show false on null values
```
