```python
import numpy as np
marks=np.array([78,92,65,88,90,45,72,55])
print(marks)
a=np.sum(marks)   #print sum
print(a)
b=np.average(marks)   #print average
print(b)
c=np.min(marks)    #print minimum
print(c)
d=np.max(marks)   #print maximum
print(d)
print(marks[marks>75])   #print above 75
print(marks[marks<50])   #print below 50
e=np.sort(marks)         #rearrange the marks with sort
print(e)
```

PANDAS as mainly two types (series and data frame)

IMPORT:-- import pandas as pd

SERIES:-- Series is an one dimensional labeled array

```python
import pandas as pd
marks=pd.Series([1,2,3,4,5]) #to used find the index
print(marks)
```

```python
import pandas as pd
marks=pd.Series([10,20,30,40],index=["maths","physics","biology","chemistry"]) #it is change the index to usable for subject name
print(marks)
```
